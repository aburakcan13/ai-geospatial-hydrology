This notebook demonstrates automated threshold alerting, inundation area calculation (km²), and interactive GeoJSON export for decision-makers.

```python
import numpy as np

def calculate_inundation_statistics(water_mask: np.ndarray, pixel_resolution_m: float = 10.0, alert_threshold_km2: float = 5.0):
    """Calculates surface water extent and triggers operational risk alerts."""
    water_pixel_count = np.sum(water_mask > 0.5)
    pixel_area_m2 = pixel_resolution_m * pixel_resolution_m
    area_km2 = (water_pixel_count * pixel_area_m2) / 1e6
    
    alert_status = "CRITICAL_RISK" if area_km2 >= alert_threshold_km2 else "NORMAL_STATUS"
    return {
        "water_pixels": int(water_pixel_count),
        "surface_area_km2": round(float(area_km2), 4),
        "alert_level": alert_status
    }

# Simulation test
dummy_mask = (np.random.rand(512, 512) > 0.8).astype(np.float32)
stats = calculate_inundation_statistics(dummy_mask, pixel_resolution_m=10.0, alert_threshold_km2=5.0)
print(f"Inundation Statistics: {stats}")
```

# Chapter 2: Multispectral Indices (MNDWI, NDWI, AWEIsh) & Thresholding
**AI-Driven Geospatial Hydrology (Springer Nature - Studies in Big Data)**

This notebook demonstrates automated calculation of water indices from Sentinel-2 bands, Otsu automatic thresholding, and shadow/cloud differentiation.

```python
import numpy as np
import matplotlib.pyplot as plt

def compute_mndwi(green_band: np.ndarray, swir_band: np.ndarray) -> np.ndarray:
    """Calculates Modified Normalized Difference Water Index (Xu, 2006)."""
    numerator = green_band - swir_band
    denominator = green_band + swir_band + 1e-6
    return numerator / denominator

# Sentetik örnek bant verisi
b3_green = np.random.uniform(0.05, 0.3, (128, 128))
b11_swir = np.random.uniform(0.01, 0.2, (128, 128))
mndwi_map = compute_mndwi(b3_green, b11_swir)
print(f"MNDWI hesaplandı. Ortalama değer: {np.mean(mndwi_map):.3f}")
```
